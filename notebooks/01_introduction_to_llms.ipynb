```python
# Write a Python function to take a list of numbers and return the count of the number of prime numbers in the list.

def check_prime(num: int):
    n = int(np.sqrt(num))
    flag = True
    for i in range(2,n+1):
        if (num%i) == 0:
            print(f'he {num} {i} : {num%i}')
            flag = False
            break
    return flag

check_prime(17)
```
